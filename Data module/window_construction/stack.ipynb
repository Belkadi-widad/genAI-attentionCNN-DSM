```python
import numpy as np
import os
```

```python
path = "./inputs/organisme/"

def stack_npy_files(path):
    files = [f for f in os.listdir(path) if f.endswith(".npy")]
    
    print("Channel order (unsorted):")
    for i, f in enumerate(files):
        print(i, "→", f)

    arrays = []
    for f in files:
        arr = np.load(os.path.join(path, f))  # load (N, H, W)
        arrays.append(arr)

    # Stack along the channel axis → (N, H, W, C)
    stacked = np.concatenate(arrays, axis=3)

    return stacked

organisme = stack_npy_files(path)
print(organisme.shape)
```

```python
np.save("./inputs_stacked/organisme.npy", organisme)
```

```python
climat = stack_npy_files("./inputs/climat/")
```

```python
np.save("./inputs_stacked/climat.npy", climat)
```

```python
relief = stack_npy_files("./inputs/relief/")
print(relief.shape)
```

```python
np.save("./inputs_stacked/relief.npy", relief)
```
