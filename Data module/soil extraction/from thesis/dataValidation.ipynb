```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
df1 = pd.read_csv('../../results/Splined/full_splined.csv') 
df2 = pd.read_csv('../../results/Agentic/results_cleaned_final.csv')

merged = pd.merge(df1, df2, on=['X', 'Y' , 'UD' , "LD"], suffixes=('_1', '_2'))
columns_to_compare = ['pH', 'CE' , "CEC" , "A%" , "L%" , "S%" , "OM(%)"]
results = []
for col in columns_to_compare:
    y_true = merged[f"{col}_1"]
    y_pred = merged[f"{col}_2"]
    # Remove rows where either y_true or y_pred is NaN
    mask = (~y_true.isna()) & (~y_pred.isna())
    y_true_clean = y_true[mask]
    y_pred_clean = y_pred[mask]

    if len(y_true_clean) == 0:
        print(f"No valid data for {col}, skipping.")
        continue

    mae = mean_absolute_error(y_true_clean, y_pred_clean)
    mse = mean_squared_error(y_true_clean, y_pred_clean)
    rmse = np.sqrt(mse)

    results.append({
        'column': col,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse
    })

results_df = pd.DataFrame(results)
print(results_df)
```
